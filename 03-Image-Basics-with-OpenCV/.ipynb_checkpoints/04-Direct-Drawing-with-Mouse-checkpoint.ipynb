# Direct Drawing with Mouse

## NOTE: THESE SHOULD ALL BE RUN AS A .py SCRIPT. View video for full details.

### Import required libraries

```python
import cv2
import numpy as np
```

### SCRIPT 1: Connecting a Function for Drawing

```python
# Defined funciton with specified parameters
def draw_circle(event, x, y, flags, param):
    if event == cv2.EVENT_LBUTTONDOWN: # using left button click
        
        # Draw circle
        cv2.circle(img=img, center=(x,y), radius=50, color=(255,255,255), thickness=-1)

# Create an blank image
img = np.zeros((512,512,3), np.uint8)

# Create a window with a name, "My Drawing"
cv2.namedWindow(winname='My Drawing')

# Whenever the user performs a mouse action inside the My Drawing window, call the draw_circle function.
cv2.setMouseCallback('My Drawing', draw_circle)

# Run forever until break using Esc key on keyboard
while True:

    cv2.imshow('My Drawing', img)
    
    # If we've waited at least 1 ms AND we've pressed the Esc
    if cv2.waitKey(20) & 0xFF == 27:
        break

# Once script is done, its usually good practice to call this line
# It closes all windows (just in case you have multiple windows called)
cv2.destroyAllWindows()
```

## SCRIPT 2: Adding Functionality with Event Choices

---------

```python
# Defined funciton with specified parameters
def draw_circle(event, x, y, flags, param):
    if event == cv2.EVENT_LBUTTONDOWN: # using left button click
        cv2.circle(img=img, center=(x,y), radius=50, color=(255,255,255), thickness=-1)
    elif event == cv2.EVENT_RBUTTONDOWN: # using right button click
        cv2.circle(img=img, center=(x,y), radius=50, color=(255,0,255), thickness=-1)

# Create an blank image
img = np.zeros((512,512,3), np.uint8)
# Create a window with a name, "My Drawing"
cv2.namedWindow(winname='My Drawing')
# Whenever the user performs a mouse action inside the My Drawing window, call the draw_circle function.
cv2.setMouseCallback('My Drawing', draw_circle)

# Run forever until break using Esc key on keyboard
while True:
    cv2.imshow('My Drawing', img)
    # If we've waited at least 1 ms AND we've pressed the Esc
    if cv2.waitKey(20) & 0xFF == 27:
        break

# Once script is done, its usually good practice to call this line
# It closes all windows (just in case you have multiple windows called)
cv2.destroyAllWindows()
```

# SCRIPT 3: Dragging with Mouse

This function can use not only for rectangle but also other irregular shapes.

```python
# Initiate variables
drawing = False  # ensure user is currently drawing or not
ix, iy = -1, -1  # '-1' means no starting point 

# Defined funciton with specified parameters
def draw_rectangle(event, x, y, flags, params):
    global drawing, ix, iy # use variables outside the function

    # When we press click:
    if event == cv2.EVENT_LBUTTONDOWN:
        drawing = True # regard as start drawing
        ix, iy = x, y  # get current pointer location

    # When we move the mouse:
    elif event == cv2.EVENT_MOUSEMOVE:
        if drawing == True:
            # Draw rectangle from previous pointer position to current
            cv2.rectangle(img=img, pt1=(ix,iy), pt2=(x,y), color=(0,255,0), thickness=-1)

    elif event == cv2.EVENT_LBUTTONUP:
        drawing = False
        cv2.rectangle(img=img, pt1=(ix,iy), pt2=(x,y), color=(0,255,0), thickness=-1)


# Create an blank image
img = np.zeros((512,512,3), np.uint8)
# Create a window with a name, "My Drawing"
cv2.namedWindow(winname='My Drawing')
# Whenever the user performs a mouse action inside the My Drawing window, call the draw_circle function.
cv2.setMouseCallback('My Drawing', draw_rectangle)

# Run forever until break using Esc key on keyboard
while True:
    cv2.imshow('My Drawing', img)
    # If we've waited at least 1 ms AND we've pressed the Esc
    if cv2.waitKey(20) & 0xFF == 27:
        break

# Once script is done, its usually good practice to call this line
# It closes all windows (just in case you have multiple windows called)
cv2.destroyAllWindows()
```
