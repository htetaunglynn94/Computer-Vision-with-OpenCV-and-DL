## Convenience Functions

Quick function for loading the puppy image.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
def load_img():
    """
    Load and convert image to 'BGR' TO 'RGB'. Return imgae.
    """
    img = cv2.imread('data/bricks.jpg').astype(np.float32) / 255
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    return img
```

```python
def display_img(img):
    """
    Display image.
    """
    plt.figure(figsize=(12,10))
    plt.imshow(img)
```

```python
img = load_img()
display_img(img)
```

-------

### Gamma Correction : Practical Effect of Increasing Brightness

`np.power(img, gamma)`  $\Rightarrow \displaystyle \boxed {\text{output pixel} = (\text{input pixel})^\gamma}$

```python
# Increasing brightness
gamma = 1/4 # 0.25
effected_img = np.power(img, gamma)
display_img(effected_img)
```

```python
# Decreasing brightness
img = load_img()
gamma = 2 
effected_img = np.power(img, gamma)
display_img(effected_img)
```

### Low Pass Filter with a 2D Convolution

A fitlering operation known as 2D convolution can be used to create a low-pass filter. Make sure to view the video for an explanation of what's happening in the code below.

```python
img.dtype
```

```python
img = load_img()
# Convert normalized image (0–1) to uint8 (0–255)
img = (img * 255).clip(0, 255).astype(np.uint8) # .clip(0, 255) => limit the range
font = cv2.FONT_HERSHEY_COMPLEX

# Add text
cv2.putText(img = img,
            text = 'bricks',
            org = (10, 600),  
            fontFace = font,
            fontScale = 10,
            color = (255,0,0), 
            thickness = 1)

display_img(img)
```

### Create the Kernel

```python
kernel = np.ones(shape=(5,5), dtype=np.float32) / 25  # 1/25 = 0.04
kernel
```

The reason why divided by 25 is that 25 elements contain inside the kernel. This kernel is used for `blurred (smoothed)` image.

```python
dst = cv2.filter2D(src = img,          # input image
                   ddepth = -1,        # keep the same output depth (data type) as the input.
                   kernel = kernel)    # apply (5 x 5) averaging matrix
display_img(dst)
```

## Averaging

```python
img = load_img()
# Convert normalized image (0–1) to uint8 (0–255)
img = (img * 255).clip(0, 255).astype(np.uint8) # .clip(0, 255) => limit the range
font = cv2.FONT_HERSHEY_COMPLEX
# Add text
cv2.putText(img = img,
            text = 'bricks',
            org = (10, 600),  
            fontFace = font,
            fontScale = 10,
            color = (255,0,0), 
            thickness = 4)
display_img(img)
```

```python
blurred_img = cv2.blur(src=img, ksize=(5,5)) # same with "cv2.filter2D()"" function
display_img(blurred_img)
```

## Gaussian Blurring

```python
img = load_img()
# Convert normalized image (0–1) to uint8 (0–255)
img = (img * 255).clip(0, 255).astype(np.uint8) # .clip(0, 255) => limit the range
font = cv2.FONT_HERSHEY_COMPLEX
# Add text
cv2.putText(img = img,
            text = 'bricks',
            org = (10, 600),  
            fontFace = font,
            fontScale = 10,
            color = (255,0,0), 
            thickness = 4)
display_img(img)
```

```python
blurred_img = cv2.GaussianBlur(src = img,
                              ksize = (5,5),
                              sigmaX = 10)
display_img(blurred_img)
```

## Median Blurring

```python
img = load_img()
# Convert normalized image (0–1) to uint8 (0–255)
img = (img * 255).clip(0, 255).astype(np.uint8) # .clip(0, 255) => limit the range
font = cv2.FONT_HERSHEY_COMPLEX
# Add text
cv2.putText(img = img,
            text = 'bricks',
            org = (10, 600),  
            fontFace = font,
            fontScale = 10,
            color = (255,0,0), 
            thickness = 4)
display_img(img)
```

```python
median = cv2.medianBlur(src=img, ksize=5)
display_img(median)
```

`cv2.medianBlur()` is the best way for **edge detection**.

## Bilateral Filtering

```python
img = load_img()
# Convert normalized image (0–1) to uint8 (0–255)
img = (img * 255).clip(0, 255).astype(np.uint8) # .clip(0, 255) => limit the range
font = cv2.FONT_HERSHEY_COMPLEX
# Add text
cv2.putText(img = img,
            text = 'bricks',
            org = (10, 600),  
            fontFace = font,
            fontScale = 10,
            color = (255,0,0), 
            thickness = 4)
display_img(img)
```

```python
blur = cv2.bilateralFilter(src=img, d=9, sigmaColor=75, sigmaSpace=75)
display_img(blur)
```
