# Gradients

Understanding gradients will allow us to eventually understand edge detection which we will use later on, since its an important aspect of object detection in general.
Make sure to watch the lecture video for the full overview!

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
def display_img(img):
    fig = plt.figure(figsize=(12,10))
    plt.imshow(img,cmap='gray')
```

```python
img = cv2.imread('data/sudoku.jpg', 0)
display_img(img)
```

```python
sobelx = cv2.Sobel(src=img, # input image
                   ddepth=cv2.CV_64F, # stores the result as 64-bit floating-point values
                   dx=1, # calculates the first derivative in the x-direction
                   dy=0, # does not calculate a derivative in the y-direction
                   ksize=5) # uses a (5 x 5) Sobel kernel to estimate the gradient
display_img(sobelx)
```

```python
sobely = cv2.Sobel(src=img, # input image
                   ddepth=cv2.CV_64F, # stores the result as 64-bit floating-point values
                   dx=0, # does not calculate a derivative in the x-direction 
                   dy=1, # calculates the first derivative in the y-direction
                   ksize=5) # uses a (5 x 5) Sobel kernel to estimate the gradient
display_img(sobely)
```

```python
laplacian = cv2.Laplacian(src=img, ddepth=cv2.CV_64F)
display_img(laplacian)
```

## Combining Previous Ideas

Let's play around with these images with some of the other ideas we've already seen!

### Blending Images

```python
blended = cv2.addWeighted(src1=sobelx, alpha=0.5, src2=sobely, beta=0.5, gamma=0)
display_img(blended)
```

```python
blended.shape
```

### Morphological Operators

```python
kernel = np.ones((4,4), np.uint8)
gradient = cv2.morphologyEx(blended, cv2.MORPH_GRADIENT, kernel)
display_img(gradient)
```

Try it on the laplacian result!

```python
kernel = np.ones((3,3), np.uint8)
gradient = cv2.morphologyEx(blended, cv2.MORPH_GRADIENT, kernel)
display_img(gradient)
```

### Thresholds

```python
ret, thresh1 = cv2.threshold(src=img,                 # input image: should be gray-scaled
                             thresh=100,              # compares every pixel against 127
                             maxval=255,              # all pixels > 127 will be white color
                             type=cv2.THRESH_BINARY)  # thresholding method to use 
# Therefore, the grayscale image becomes a black-and-white binary image.
# 'ret' includes threshold value and thresh1 includes resulting thresholded image

display_img(thresh1)
```

```python
ret, thresh2 = cv2.threshold(src=gradient,                # input image: should be gray-scaled
                             thresh=200,                  # compares every pixel against 127
                             maxval=255,                  # all pixels > 127 will be white color
                             type=cv2.THRESH_BINARY_INV)  # thresholding method to use 
# Therefore, the grayscale image becomes a black-and-white binary image.
# 'ret' includes threshold value and thresh1 includes resulting thresholded image

display_img(thresh2)
```

```python
ret, thresh3 = cv2.threshold(src=blended,                 # input image: should be gray-scaled
                             thresh=100,                  # compares every pixel against 127
                             maxval=255,                  # all pixels > 127 will be white color
                             type=cv2.THRESH_BINARY_INV)  # thresholding method to use 
# Therefore, the grayscale image becomes a black-and-white binary image.
# 'ret' includes threshold value and thresh1 includes resulting thresholded image

display_img(thresh3)
```
