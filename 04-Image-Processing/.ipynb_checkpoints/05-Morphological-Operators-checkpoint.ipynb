# Morphological Operators

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
def load_img():
    blank_img = np.zeros((600,600), dtype='uint8')
    font = cv2.FONT_HERSHEY_SIMPLEX
    cv2.putText(blank_img, text='ABCDE', org=(50,300), fontFace=font, fontScale=5, color=(255,255,255), thickness=25, lineType=cv2.LINE_AA)
    return blank_img
```

```python
def display_img(img):
    plt.figure(figsize=(12,10))
    plt.imshow(img, cmap='gray')
```

```python
img = load_img()
display_img(img)
```

## Erosion

Erodes away boundaries of foreground objects. Works best when foreground is light color (preferrably white) and background is dark.

```python
kernel = np.ones((5,5), np.uint8)
kernel
```

```python
# Shrinks white objects
erosion1 = cv2.erode(src=img, kernel=kernel, iterations=1)
display_img(erosion1)
```

```python
img = load_img()
kernel = np.ones((5,5), np.uint8)
erosion2 = cv2.erode(src=img, kernel=kernel, iterations=3)
display_img(erosion2)
```

## Opening
Opening is erosion followed by dilation. Useful in removing background noise!

```python
img = load_img()
```

```python
white_noise = np.random.randint(0, 2, size=(600,600), dtype=np.uint8) # create matrix including 0 and 1 only
```

```python
white_noise = white_noise * 255 # convert matrix including 0 and 255 only
white_noise
```

```python
white_noise.shape, img.shape
```

```python
noise_img = white_noise + img
noise_img
```

```python
display_img(noise_img)
```

```python
# Opening is erosion followed by dilation
opening = cv2.morphologyEx(src=noise_img, op=cv2.MORPH_OPEN, kernel=kernel)
display_img(opening)
```

### Closing
 Useful in removing noise from foreground objects, such as black dots on top of the white text.

```python
img = load_img()
```

```python
black_noise = np.random.randint(0, 2, size=(600,600))
black_noise
```

```python
black_noise = (black_noise * (-255)).astype(np.uint8)
```

```python
black_noise_img = img + black_noise
```

```python
black_noise_img
```

```python
display_img(black_noise_img)
```

```python
black_noise_img[black_noise_img == -255] = 0
display_img(black_noise_img)
```

```python
# Closing is dilation followed by erosion
closing = cv2.morphologyEx(src=black_noise_img, op=cv2.MORPH_CLOSE, kernel=kernel)
```

```python
display_img(closing)
```

## Morphological Gradient

Difference between dilation and erosion of an image.

```python
img = load_img()
display_img(img)
```

```python
gradient = cv2.morphologyEx(src=img, op=cv2.MORPH_GRADIENT, kernel=kernel)
display_img(gradient)
```
