# Images and Numpy

Numpy can read in certain files types, this includes images stored as arrays. In this quick lecture, we'll quickly cover how you can work with images in NumPy. **Keep in mind, we will mainly be using OpenCV to open and view images, but later on in the Deep Learning Section we will come back to NumPy and Matplotlib. **

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

from PIL import Image
```

```python
pic = Image.open('data/00-puppy.jpg')
pic
```

```python
type(pic)
```

```python
pic_arr = np.asarray(pic)
pic_arr.shape
```

```python
plt.imshow(pic_arr)
plt.show()
```

```python
def compare_images(a, b, c, cmap=None):
    fig, x = plt.subplots(1,3,figsize=(10,12))

    x[0].imshow(a[:,:,0], cmap=cmap)
    x[1].imshow(b[:,:,1], cmap=cmap)
    x[2].imshow(c[:,:,2], cmap=cmap)
    
    x[0].set_title("RED CHANNEL")
    x[1].set_title("GREEN CHANNEL")
    x[2].set_title("BLUE CHANNEL")
    
    x[0].axis('off')
    x[1].axis('off')
    x[2].axis('off')
    
    plt.tight_layout()
    plt.show()
```

```python
compare_images(pic_arr, pic_arr, pic_arr, cmap='gray')
```

**RED | GREEN | BLUE**  
- All of the value are between 0 to 255
- The brightest areas (white areas) have the most dedicated channel in each image
