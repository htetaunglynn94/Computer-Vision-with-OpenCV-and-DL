# Grid Detection

```python
import cv2
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
flat_chess = cv2.imread('data/flat_chessboard.png')
```

```python
plt.imshow(flat_chess, cmap='gray')
```

```python
found, corners = cv2.findChessboardCorners(image=flat_chess, 
                                           patternSize=(7,7))
# NOTE: We use (7 x 7) instead (8 x 8) because the outer margin won't be considered.
# found: boolean
# corners: numbers
```

```python
if found:
    print('OpenCV was able to find the corners')
else:
    print("OpenCV did not find corners. Double check your patternSize.")
```

```python
corners.shape
```

```python
flat_chess_copy = flat_chess.copy()
cv2.drawChessboardCorners(flat_chess_copy, (7, 7), corners, found)
```

```python
plt.imshow(flat_chess_copy)
```

# Circle Based Grids

```python
dots = cv2.imread('data/dot_grid.png')
```

```python
plt.imshow(dots)
```

```python
found, corners = cv2.findCirclesGrid(image=dots, patternSize=(10,10), flags=cv2.CALIB_CB_SYMMETRIC_GRID)
```

```python
found
```

```python
dbg_image_circles = dots.copy()
cv2.drawChessboardCorners(dbg_image_circles, (10, 10), corners, found)
```

```python
plt.imshow(dbg_image_circles)
```
